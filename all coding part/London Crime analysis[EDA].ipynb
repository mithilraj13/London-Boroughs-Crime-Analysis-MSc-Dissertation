```python
import pandas as pd

crime = pd.read_csv("data/cleaned_crime_long.csv", parse_dates=["Date"])

print(crime.shape)
print(crime.dtypes)
```

```python
top_boroughs = crime.groupby("Borough")["CrimeCount"].sum().sort_values(ascending=False)

print(top_boroughs.head(10))
```

```python
top_categories = crime.groupby("MajorCategory")["CrimeCount"].sum().sort_values(ascending=False)

print(top_categories)
```

```python
yearly = crime.groupby("Year")["CrimeCount"].sum()

print(yearly)
```

```python
seasonal = crime.groupby("Month")["CrimeCount"].mean()

print(seasonal)
```

```python
months_per_borough = crime.groupby("Borough")["Date"].nunique()

print(months_per_borough.describe())
print(months_per_borough.unique())
```
